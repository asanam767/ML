```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x='weather_condition', hue='most_severe_injury', data=df)
plt.title('Accidents by Weather Condition and Injury Severity')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x='roadway_surface_cond', hue='crash_type', data=df)
plt.title('Accidents by Road Surface Condition')
plt.xticks(rotation=45)
plt.show()
```

```python
df['crash_month'].value_counts().sort_index().plot(kind='line', title='Accidents by Month')
```

```python
df['crash_day_of_week'].value_counts().sort_index().plot(kind='bar', title='Accidents by Day of Week')
```

```python
df['crash_hour'].value_counts().sort_index().plot(kind='bar', title='Accidents by Hour of the Day')
```

```python
corr = df[['injuries_total', 'num_units', 'injuries_fatal']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
df['prim_contributory_cause'].value_counts().plot(kind='bar', title='Primary Contributory Causes')
```

```python
injury_columns = ['injuries_fatal', 'injuries_incapacitating', 'injuries_non_incapacitating']
df[injury_columns].sum().plot(kind='bar', title='Injuries by Type')
```

```python
sns.countplot(x='trafficway_type', hue='crash_type', data=df)
plt.title('Accidents by Roadway Type')
plt.xticks(rotation=45)
plt.show()
```

```python
pivot_table = df.pivot_table(index='crash_hour', columns='crash_day_of_week', values='crash_date', aggfunc='count')
sns.heatmap(pivot_table, cmap='YlOrRd')
plt.title('Accidents by Hour and Day of Week')
plt.show()
```

```python
df['crash_type'].value_counts().plot(kind='pie', autopct='%1.1f%%', title='Crash Types')
plt.show()
```
